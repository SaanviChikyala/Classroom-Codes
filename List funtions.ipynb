```python
numbers=[]
for i in range(20):
    num=int(input("Enter value:"))
    numbers.append(num)
print("List:", numbers)

print("Similar elements and their index values:")
for i in range(20):
    if numbers.count(numbers[i])>1:
        print(numbers[i],"->", end="")
    for j in range (20):
        if numbers[j]==numbers[i]:
            print(j,end="")
            print()
            
# Count even and odd
even=0
odd=0
for num in numbers:
    if num%2==0:
        even+=1
    else:
        odd+=1
print("Even values:",even)
print("Odd values:", odd)

for num in numbers:
    if num>0:
        positive+=1
    elif num<0:
        negative+=1
print("Positive values:", positive)
print("Negative values:", negative)

print("Length:", len(numbers))
print("Ascending:", sorted(numbers))
print("Descending:", numbers.sort(reverse=True))
```
