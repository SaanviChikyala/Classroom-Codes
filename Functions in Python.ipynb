```python
def evenodd(x:int):
    if (x%2==0):
        return "Even"
    else:
        return "Odd"
print(evenodd(8))
print(evenodd(5))
```

```python
def calc(x,y,z):
    sum1=x+y+z
    product=x*y*z
    return sum1,product
a=int(input("Enter a:"))
b=int(input("Enter b:"))
c=int(input("Enter c:"))
s,p=calc(a,b,c)
print(s,p)
```
