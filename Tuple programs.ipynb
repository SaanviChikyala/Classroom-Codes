```python
num=(10,20,30,40,50)
ele=int(input("Enter element:"))
if ele in num:
    print("Index of element:",num.index(ele))
else:
    print("Element not found")
```

```python
num=[(10,20,40),(40,50,60),(70,80,90)]
newlist=[]
for tup in num:
    newtuple=tup[:-1]+(100,)
    newlist.append(newtuple)
print("Original list:",num)
print("Updated list:",newlist)
```
