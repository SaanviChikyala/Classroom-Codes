```python
# To reverse any 2 or 3 digit number
num=list(input("Enter number:"))
y=len(num)
for i in range(0,y-1):
    i+=1
    x=num[i]
    num[i]=num[y-1-i]
    num[y-1-i]=x
    reverse_num="".join(num)
print(reverse_num)
```

```python
# To reverse any number
num=list(input("Enter number:"))
y=len(num)
for i in range(y//2):
    x=num[i]
    num[i]=num[y-1-i]
    num[y-1-i]=x
    reverse_num="".join(num)
print(reverse_num)
```

```python
# To find factorial of any given number
num=int(input("Enter number:"))
def fact(num):
    result=1
    while num>1:
        result*=num
        num-= 1
    return result
print("The factorial of given number is",fact(num))
```

```python
# To find whether a given year is a leap year or not
year=int(input("Enter year:"))
if year%4==0:
    print("It's a leap year")
else:
    print("It's not a leap year")
```
