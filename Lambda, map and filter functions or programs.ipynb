```python
def square(num):
    return num**2
list1=[2,3,4,5,6,7,42,23,54,67,99]
list(map(square,list1))
```

```python
list(filter(lambda n:n%2==0,list1))
```

```python
def even(num):
    if num%2==0:
        return num
    else:
        return 0
list1=[2,34,21,36,76,89,3,4,6,9]
list(filter(even,list1))
```
